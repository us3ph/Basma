```python
# Basma v2 on the NTIRE 2026 Robust AI-Generated Image Detection public test (2,500 labelled images)
import os, glob, subprocess
subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/us3ph/Basma.git', '/tmp/Basma'], check=True)
os.chdir('/tmp/Basma'); print(subprocess.run(['git', 'log', '--oneline', '-1'], capture_output=True, text=True).stdout)
print(subprocess.run(['nvidia-smi', '--query-gpu=name,memory.total', '--format=csv'], capture_output=True, text=True).stdout)
print(subprocess.run('pip install -q -r requirements.txt 2>&1 | tail -1', shell=True, capture_output=True, text=True).stdout)
# the v2 model = output of the basma-train-v2 kernel (kernel outputs mount at /kaggle/input/notebooks/<owner>/<slug>/)
tgz = sorted(glob.glob('/kaggle/input/**/basma_model.tgz', recursive=True)); print(tgz); assert tgz, 'basma_model.tgz not mounted'
subprocess.run(['tar', 'xzf', tgz[0], '-C', '/tmp/Basma'], check=True); print(os.listdir('runs/kaggle_v2/best'))
from huggingface_hub import hf_hub_download
R = 'deepfakesMSU/NTIRE-RobustAIGenDetection-test-public'
z = hf_hub_download(R, 'test_images.zip', repo_type='dataset', local_dir='/tmp/ntire')
hf_hub_download(R, 'test_labels.csv', repo_type='dataset', local_dir='/tmp/ntire')
subprocess.run(['unzip', '-q', '-o', z, '-d', '/tmp/ntire/img'], check=True)
print(sum(1 for _ in glob.iglob('/tmp/ntire/img/**/*.jpg', recursive=True)), 'images unpacked')
```
