```python
import os
for root, dirs, files in os.walk('/kaggle/input'):
    depth = root.count('/') - 2
    if depth <= 2:
        print(root, '| dirs:', sorted(dirs)[:8], '| files:', sorted(files)[:6], '...' if len(files) > 6 else '')
    if depth >= 2:
        dirs[:] = []
import glob; print('MANIFESTS', sorted(glob.glob('/kaggle/input/**/manifest_*.jsonl', recursive=True)))
```
